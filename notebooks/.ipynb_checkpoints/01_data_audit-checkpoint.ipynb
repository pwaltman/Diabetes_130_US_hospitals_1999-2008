# Missingness Discussion
* It __appears__ that there is some structure in the missingness of the weight data with respect to the **`readmitted`** column. However, there appears to be little-to-no relationship with the **`time_in_hospital`** variable.
* The original authors ignored weight and payer_code because they were too sparse, and payer code wasn't considered to have any predictive value.
* The hypothesis that payer_code missingness could serve as a proxy for lack of health insurance and lack of access to preventative/proactive medical care was **NOT** demonstrated as evaluations showed it to have little significance and little effect.

# What are we predicting?
Ideally, the factors that predict whether someone will be readmitted in fewer than 30 days. Alternatively, we may seek to determine whether someone was readmitted at all.

## The Three Readmitted classes
1. **`NO:`** The patient wasn't readmitted again after this visit
2. **`>30:`** The patient was readmitted again, but the gap between visits was more than 30 days (no upper bound)
3. **`<30:`** The patient was readmitted again, but the gap between visits was less than 30 days

**NOTE:** This column doesn't include hospitalizations outside of this network, nor does it include outpatient visits.

## How common is the <30 day readmission?
1. Overall, it occurs 11.16% of the time.
2. Per patient, it occurs 12.36% of the time. 

## Obvious patterns
1. the age range [20-30) is much more likely to have a readmission less than 30 days than any other group
2. the age range [0-10) is much more likely to have a "NO" readmission than any other age range.

### Conclusions about the number of encounters per patient:
* The vast majority of patients only have one encounter.
* However, there is a small cohort of "frequent flyers" who skew the distribution.
* **One potential avenue of comparison is to compare the frequent flyers to the others to determine what distinguishes them**

## Potential Data Leakage:
The encounter ID isn't randomized, and increases as the number of hospitalizations takes place, so that it can be used to infer an ordering of the visits for a given patient, and be used by a learner to learn that a given patient will be hospitalized. For example, if two constructed features are added which indicate the number of encounters for a given patient BEFORE a test/train split is performed, that variable will provide information to the learner if the encounters are split between the test/train set.

```python
import seaborn as sns
import matplotlib.pyplot as plt
from statannotations.Annotator import Annotator

def violin_compare(
    column,
    df=df,
    group="readmitted",
    pairs=[("NO", "<30"), ("NO", ">30")]
):
    ax = sns.violinplot(
        data=df,
        x="readmitted",
        y=column
    )

    annotator = Annotator(
        ax,
        pairs,
        data=df,
        x="readmitted",
        y=column
    )

    annotator.configure(
        test="Mann-Whitney",
        text_format="simple",
        loc="inside"
    )

    annotator.apply_and_annotate()

    ax.set_xlabel("Readmitted")
    ax.set_ylabel(column)
    ax.set_title(f"{column} by readmission status")

    plt.show()
```

# SUMMARY
* There exists substantial missingness in the weight, payer_code & medical_specialty variables
    * Substantial missingness exists in weight (97%), payer_code (40%), and medical_specialty (47%).
    * The original study excluded weight because of extreme missingness and payer code because of missingness and its perceived lack of relevance to readmission.
    * Medical specialty was retained, with missingness treated as a separate category. Weight missingness may reflect historical differences in electronic health record practices.
* Most the variables are categorical. The variables tht contain numeric values primarily describe the number of medical events (inpatient, outpatient, emergency in the past year
* the large size of the data set allows for highly signficant p-values for correlations that often have little effect
* there is some temporal leakage for the numeric events for patients that have had multiple encounters. for this reason, randomization and sampling of the data should be done per patient, rather than per encounter
* \>51K of the patients only have a single enounter, while a small minority are repeat patients (so-called frequent flyers - those with \>9 visits)
    * It is known that some of the patients' encounters aren't captured by the data set.
    * However, it is assumed that it is unlikely that the bulk of these \>51k patients are a part of the frequent flyers
    * Additionally, some of the repeat patients may only have infrequent encounters, i.e. Once every 1-2 years
* One goal of the project is to determine the viability of distinguishing between the singleton and the frequent flyer patients.
* A secondary goal of the project is to determine why a frequent flyer patient may need more care in some circumstances than others.
